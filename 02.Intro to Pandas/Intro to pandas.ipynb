# Creating a dataframe from an array

## Option 1

```python
# creating an array
data = np.array([[1,4], [2,5], [3,6]])
```

```python
# creating a dataframe
df = pd.DataFrame(data,
             index = ["row1","row2", "row3"],
             columns=["col1", "col2"])
```

```python
# showing the data frame
df
```

## Option 2

```python
# creating an array with list shape
data2 = [[1,4], [2,5], [3,6]]
```

```python
# creating a dataframe
df2 = pd.DataFrame(data2,
                  index = ["row1","row2", "row3"],
                  columns=["col1", "col2"])
```

```python
# showing the dataframe
df2
```

# Creating a DataFrame from a dictionary

```python
# list used for this example
states = ["Texas", "Georgia", "Michigan","Florida"]

population = [12345678, 59874635, 8749587, 94536125]
```

```python
# sorting lists within a dictionary
dict_states = {'States':states, 'Population':population}
```

```python
# Creating the dataframe
df_states = pd.DataFrame(dict_states)
```

```python
# Showing the dataframe
df_states
```

# Adding a new column

## assign()

```python
# create random numbers
score1 = np.random.randint(0,101, size=(len(df_exams)))
score2 = np.random.randint(0, 101, size=(len(df_exams)))
```

```python
# create series 
series1 = pd.Series(score1)
series2 = pd.Series(score2)
```

```python
series2
```

```python
df_exams.assign(score1=series1, score2=series2)
```

```python
df_exams = df_exams.assign(score1=series1, score2=series2)
df_exams
```

## insert()

```python
# using insert() to add column
df_exams.insert(2, "test", series1)
```

```python
df_exams
```
